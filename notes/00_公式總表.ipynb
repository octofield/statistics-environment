# 公式總表

依講義 02–06 整理。每個公式下方附 Python 寫法；記號：母體 $N, \mu, \sigma$，樣本 $n, \bar{x}, s$。

## 一、敘述統計（03a / 03b / 06b）

| 統計量 | 公式 | Python |
|---|---|---|
| 平均數 | $\mu = \frac{1}{N}\sum x_i$、$\bar{x} = \frac{1}{n}\sum x_i$ | `statistics.mean(x)` |
| 中位數 | 排序後最中間（偶數個取中間兩個平均） | `statistics.median(x)` |
| 眾數 | 出現最多次的值；分組資料取眾數組的組中點 | `statistics.multimode(x)` |
| 幾何平均數 | $\sqrt[n]{x_1 x_2 \cdots x_n}$ | `statistics.geometric_mean(x)` |
| 全距 | 最大值 − 最小值 | `max(x) - min(x)` |
| 平均絕對離差 MAD | $\frac{1}{N}\sum \lvert x_i - \mu \rvert$ | `statistics.mean([abs(v - m) for v in x])` |
| 母體變異數 | $\sigma^2 = \frac{1}{N}\sum (x_i - \mu)^2$ | `statistics.pvariance(x)` |
| 樣本變異數 | $s^2 = \frac{1}{n-1}\sum (x_i - \bar{x})^2$ | `statistics.variance(x)` |
| 標準差 | $\sigma = \sqrt{\sigma^2}$、$s = \sqrt{s^2}$ | `statistics.pstdev(x)`、`statistics.stdev(x)` |
| 變異係數 | $\text{CV} = \sigma / \mu$（樣本 $s / \bar{x}$） | `statistics.stdev(x) / statistics.mean(x)` |
| z 分數 | $z = \frac{x - \mu}{\sigma}$ | `(x - mean) / sd` |
| 四分位數 | $Q_1, Q_2, Q_3$ | `statistics.quantiles(x, n=4)` |
| 百分位數 | 第 k 百分位數 | `statistics.quantiles(x, n=100)[k - 1]` |
| 四分位距 | $\text{IQR} = Q_3 - Q_1$ | `q3 - q1` |
| 離群值（1.5 IQR） | $< Q_1 - 1.5\,\text{IQR}$ 或 $> Q_3 + 1.5\,\text{IQR}$ | `sk.outliers_iqr(x)` |
| 偏態（講義） | $\frac{\frac{1}{n}\sum (x_i - \bar{x})^3}{s^3}$；> 0 右偏、< 0 左偏 | 見 `sk.describe(x)` |
| 峰態（講義） | $\frac{\frac{1}{n}\sum (x_i - \bar{x})^4}{s^4}$ | 見 `sk.describe(x)` |
| 母體共變異數 | $\sigma_{xy} = \frac{1}{N}\sum (x_i - \mu_x)(y_i - \mu_y)$ | `sk.cov_corr(x, y)` |
| 樣本共變異數 | $s_{xy} = \frac{1}{n-1}\sum (x_i - \bar{x})(y_i - \bar{y})$ | `statistics.covariance(x, y)` |
| 相關係數 | $r = \frac{s_{xy}}{s_x s_y}$，$-1 \le r \le 1$ | `statistics.correlation(x, y)` |
| 分組平均數 | $\bar{x} \approx \frac{\sum f_i m_i}{n}$（$m_i$ 組中點、$f_i$ 次數） | `sk.grouped_stats(m, f)` |
| 分組變異數 | $s^2 \approx \frac{\sum f_i (m_i - \bar{x})^2}{n - 1}$ | `sk.grouped_stats(m, f)` |

- **Chebyshev 定理**：任何資料，至少 $1 - \frac{1}{k^2}$ 落在 $\mu \pm k\sigma$ 內（$k > 1$）→ `sk.chebyshev(x)`
- **經驗法則**：鐘形資料約 68% / 95% / 99.7% 落在 $\mu \pm 1\sigma / 2\sigma / 3\sigma$ → `sk.empirical_rule(x)`
- **偏態判斷**：平均數 > 中位數 → 傾向右偏；平均數 < 中位數 → 傾向左偏
- **Bessel 校正**：樣本變異數除以 $n - 1$，才會讓 $E[S^2] = \sigma^2$（不偏）
- pandas 的 `.var()`、`.std()` 預設是**樣本**（ddof=1）；`.quantile()` 的算法跟 `statistics.quantiles` 不同

## 二、機率（04a）

| 規則 | 公式 |
|---|---|
| 餘事件 | $\Pr(A^c) = 1 - \Pr(A)$ |
| 加法 | $\Pr(A \cup B) = \Pr(A) + \Pr(B) - \Pr(A \cap B)$ |
| 互斥 | $\Pr(A \cap B) = 0$ → $\Pr(A \cup B) = \Pr(A) + \Pr(B)$ |
| 條件機率 | $\Pr(A \mid B) = \frac{\Pr(A \cap B)}{\Pr(B)}$ |
| 乘法 | $\Pr(A \cap B) = \Pr(A \mid B)\Pr(B)$ |
| 獨立 | $\Pr(A \cap B) = \Pr(A)\Pr(B)$，等價於 $\Pr(A \mid B) = \Pr(A)$ |
| 全機率 | $\Pr(B) = \sum_i \Pr(B \mid A_i)\Pr(A_i)$（$A_i$ 互斥且涵蓋全部） |
| 貝氏定理 | $\Pr(A_i \mid B) = \frac{\Pr(B \mid A_i)\Pr(A_i)}{\sum_j \Pr(B \mid A_j)\Pr(A_j)}$ |

| 計數 | 公式 | Python |
|---|---|---|
| 階乘 | $n!$ | `math.factorial(n)` |
| 排列（取 k 個、有順序） | $\frac{n!}{(n-k)!}$ | `math.perm(n, k)` |
| 組合（取 k 個、無順序） | $\binom{n}{k} = \frac{n!}{k!(n-k)!}$ | `math.comb(n, k)` |

## 三、隨機變數（04b / 05）

| 項目 | 離散 | 連續 |
|---|---|---|
| 機率 | pmf $\Pr(X = x)$，$\sum_x \Pr(X = x) = 1$ | pdf $f(x)$，$\int f(x)\,dx = 1$；$\Pr(X = a) = 0$ |
| 累積分布 | $F(x) = \Pr(X \le x) = \sum_{t \le x} \Pr(X = t)$ | $F(x) = \int_{-\infty}^{x} f(t)\,dt$ |
| 區間機率 | $\Pr(a \le X \le b) = F(b) - F(a - 1)$（整數值） | $\Pr(a \le X \le b) = F(b) - F(a)$ |
| 期望值 | $E[X] = \sum_x x \Pr(X = x)$ | $E[X] = \int x f(x)\,dx$ |
| 變異數 | $\operatorname{Var}(X) = \sum_x (x - \mu)^2 \Pr(X = x)$ | $\operatorname{Var}(X) = \int (x - \mu)^2 f(x)\,dx$ |

- $\operatorname{Var}(X) = E[X^2] - (E[X])^2$、$\operatorname{SD}(X) = \sqrt{\operatorname{Var}(X)}$
- **線性轉換**：$E[a + bX] = a + bE[X]$、$\operatorname{Var}(a + bX) = b^2 \operatorname{Var}(X)$
- **和**：$E[X_1 + \cdots + X_n] = E[X_1] + \cdots + E[X_n]$（永遠成立）
- **獨立時**：$E[X_1 \cdots X_n] = E[X_1] \cdots E[X_n]$、$\operatorname{Var}(X_1 + \cdots + X_n) = \operatorname{Var}(X_1) + \cdots + \operatorname{Var}(X_n)$
- 自訂 pdf 求常數 $k$：解 $\int f(x)\,dx = 1$（snippet `sympdf`）；自訂機率表：`sk.Discrete(值, 機率)`

## 四、分布對照表（04c / 05 / 06a）

| 分布 | 意義 | pmf / pdf | 可能值 | $E[X]$ | $\operatorname{Var}(X)$ |
|---|---|---|---|---|---|
| 白努利 Ber(p) | 一次試驗成功與否 | $p^x(1-p)^{1-x}$ | 0, 1 | $p$ | $p(1-p)$ |
| 二項 Bi(n, p) | n 次試驗的成功次數 | $\binom{n}{x}p^x(1-p)^{n-x}$ | 0, 1, …, n | $np$ | $np(1-p)$ |
| 幾何 Geo(p) | 第幾次試驗才第一次成功 | $(1-p)^{x-1}p$ | 1, 2, … | $\frac{1}{p}$ | $\frac{1-p}{p^2}$ |
| 超幾何 Hyper(N, K, n) | 不放回抽 n 個的成功數 | $\frac{\binom{K}{x}\binom{N-K}{n-x}}{\binom{N}{n}}$ | 0, …, min(n, K) | $n\frac{K}{N}$ | $n\frac{K}{N}\left(1-\frac{K}{N}\right)\frac{N-n}{N-1}$ |
| 卜瓦松 Poi(λ) | 單位時間內發生次數 | $\frac{e^{-\lambda}\lambda^x}{x!}$ | 0, 1, 2, … | $\lambda$ | $\lambda$ |
| 均勻 Uni(a, b) | 區間內每個值一樣可能 | $\frac{1}{b-a}$ | $[a, b]$ | $\frac{a+b}{2}$ | $\frac{(b-a)^2}{12}$ |
| 指數 Exp(λ) | 等待時間（λ 為發生率） | $\lambda e^{-\lambda x}$ | $x \ge 0$ | $\frac{1}{\lambda}$ | $\frac{1}{\lambda^2}$ |
| 常態 ND(μ, σ) | 鐘形 | $\frac{1}{\sigma\sqrt{2\pi}}e^{-\frac{(x-\mu)^2}{2\sigma^2}}$ | 實數 | $\mu$ | $\sigma^2$ |

**累積分布函數**：均勻 $F(x) = \frac{x-a}{b-a}$；指數 $F(x) = 1 - e^{-\lambda x}$、$\Pr(X > x) = e^{-\lambda x}$

**scipy 寫法**（pmf / pdf / cdf / ppf 都是同樣的參數）

| 分布 | scipy | 注意 |
|---|---|---|
| Bi(n, p) | `stats.binom.cdf(k, n=n, p=p)` | |
| Geo(p) | `stats.geom.cdf(k, p=p)` | k 從 1 開始 |
| Hyper(N, K, n) | `stats.hypergeom.cdf(k, M=N, n=K, N=n)` | **名稱跟課本不同** |
| Poi(λ) | `stats.poisson.cdf(k, mu=λ)` | |
| Uni(a, b) | `stats.uniform.cdf(x, loc=a, scale=b - a)` | scale 是寬度 |
| Exp(λ) | `stats.expon.cdf(x, scale=1/λ)` | **scale = 1/λ** |
| ND(μ, σ) | `stats.norm.cdf(x, loc=μ, scale=σ)` | scale 是標準差 |

**離散分布的不等號**（X 只取整數）：$\Pr(X \le k) = F(k)$、$\Pr(X < k) = F(k-1)$、$\Pr(X \ge k) = 1 - F(k-1)$、$\Pr(X > k) = 1 - F(k)$

## 五、常態分布與近似（06a）

- **標準化**：$Z = \frac{X - \mu}{\sigma} \sim \text{ND}(0, 1)$；$\Pr(X \le x) = \Pr\left(Z \le \frac{x - \mu}{\sigma}\right)$
- 反過來：$x = \mu + z\sigma$（例：第 60 百分位數 $= \mu + z_{0.6}\,\sigma$，`stats.norm.ppf(0.6) * σ + μ`）
- 常用 z 值：$z_{0.90} = 1.2816$、$z_{0.95} = 1.6449$、$z_{0.975} = 1.9600$、$z_{0.99} = 2.3263$、$z_{0.995} = 2.5758$

**二項 → 常態**：$\text{Bi}(n, p) \approx \text{ND}\left(np, \sqrt{np(1-p)}\right)$，經驗法則（講義）：$n \ge 25$、$np > 5$、$n(1-p) > 5$

**連續校正**（$Y$ 為近似的常態）

| 二項 | 常態近似 |
|---|---|
| $\Pr(X \le b)$ | $\Pr(Y \le b + 0.5)$ |
| $\Pr(X \ge a)$ | $\Pr(Y \ge a - 0.5)$ |
| $\Pr(a \le X \le b)$ | $\Pr(a - 0.5 \le Y \le b + 0.5)$ |
| $\Pr(X = k)$ | $\Pr(k - 0.5 \le Y \le k + 0.5)$ |

`sk.normal_approx(n, p, '25 <= X <= 30')` 會同時算精確值、無校正、連續校正。

**其他近似**
- 二項 → 卜瓦松：$n \to \infty$、$p \to 0$、$\lambda = np$
- 超幾何 → 二項：$\frac{n}{N} \to 0$（抽的比例很小時，不放回 ≈ 放回），$p = \frac{K}{N}$

## 六、應用與抽樣（04b / 05 / 06a / 06b）

- **卜瓦松與指數**：每單位時間發生次數 $\sim \text{Poi}(\lambda)$ ⇔ 兩次發生的間隔時間 $\sim \text{Exp}(\lambda)$
- **指數的無記憶性**：$\Pr(X > s + t \mid X > s) = \Pr(X > t)$
- **報童模型**：$c_o = c + d - v$（多訂一單位的損失）、$c_u = r - c + s$（少訂一單位的損失），最適訂購量 $F(q^*) = \frac{c_u}{c_o + c_u}$、缺貨機率 $1 - F(q^*) = \frac{c_o}{c_o + c_u}$ → `sk.newsvendor(D, co, cu)`
- **離散需求的存貨**：服務水準 $= \Pr(X \le y)$；期望利潤逐一計算 → `sk.inventory_profit(D, price, cost)`
- **樣本平均數**：$E[\bar{X}] = \mu$、$\operatorname{Var}(\bar{X}) = \frac{\sigma^2}{n}$、$\operatorname{SD}(\bar{X}) = \frac{\sigma}{\sqrt{n}}$；母體為常態時 $\bar{X} \sim \text{ND}\left(\mu, \frac{\sigma}{\sqrt{n}}\right)$；$n$ 夠大時不論母體分布都近似常態（中央極限定理）→ `sk.sample_means(D, n)`
- **大數法則**：$n \to \infty$ 時 $\bar{X}_n \to \mu$
- **蒙地卡羅**：用大量隨機模擬估計機率或期望值，例如 $\pi \approx 4 \times \frac{\text{圓內點數}}{\text{總點數}}$ → `sk.estimate_pi(n)`

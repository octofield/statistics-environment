## 考試流程

1. 插隨身碟 → 雙擊 `1_啟動VSCode.bat`（備案：`2_啟動JupyterLab.bat`）
2. 下載老師給的考卷 ipynb（和資料檔）→ 雙擊 `5_匯入考卷.bat`（或 VS Code 裡 Ctrl+Shift+P →「執行工作」→「匯入考卷」）→ 選檔案 → 自動複製到 `exam/`（保留原檔名；資料檔一起複製）、標題下面加上 import 與設定並打開 → **先執行那一格**，再在開頭的 Name / Student ID 寫上學號姓名
3. 右上角選 Kernel：**Python 3（WPy64 那個）**（通常會自動選好）
4. 有資料檔：跟考卷放在一起（`exam/`）→ `df = sk.peek('檔名.xlsx')` → 照「讀入資料與整理」檢查、整理後再作答；每一小題用 snippet `ans` 印出作答格式，答案用 `print` 印出來
5. 作答，隨時 Ctrl+S（已開自動存檔）
6. 寫完：Ctrl+Shift+P →「執行工作」→「匯出目前的 notebook 成 HTML（先從頭重新執行）」（或雙擊 `3_匯出HTML.bat`）→ 瀏覽器檢查 → 上傳 COOL
7. 遇到問題先看 [常見問題.md](常見問題.md)；考完先關 VS Code 再安全移除隨身碟
